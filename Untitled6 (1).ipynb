```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import train_test_split, GridSearchCV
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.optimizers import Adam
from scikeras.wrappers import KerasClassifier
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, roc_curve, auc



# préparation données
df = pd.read_csv('/home/webcressontech/Documents/ECE/Deep-learning/creditcard.csv')

print(df.head())

print(df.info())

print(df['Class'].value_counts())



X = df.drop('Class', axis=1)
y = df['Class']

# Normalisation des données
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Rééquilibrage des classes avec SMOTE
smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X_scaled, y)

# Séparation en train et test (80% train, 20% test)
X_train, X_test, y_train, y_test = train_test_split(X_resampled, y_resampled, test_size=0.2, random_state=42)


print(f"Dimensions de X_train : {X_train.shape}")
print(f"Dimensions de X_test : {X_test.shape}")
print(f"Dimensions de y_train : {y_train.shape}")
print(f"Dimensions de y_test : {y_test.shape}")



# Créer un modèle de réseau de neurones
model = Sequential([
    Dense(128, activation='relu', input_shape=(X_train.shape[1],)),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')  # Sortie binaire (frauduleuse ou non)
])

# Compiler le modèle
model.compile(optimizer=Adam(), loss='binary_crossentropy', metrics=['accuracy'])

history = model.fit(X_train, y_train, epochs=10, batch_size=64, validation_data=(X_test, y_test))

model.summary()


# Prédire les résultats sur les données de test
y_pred = model.predict(X_test)
y_pred_binary = (y_pred > 0.5).astype(int)  # Convertir les probabilités en classes binaires

# Calculer la matrice de confusion
cm = confusion_matrix(y_test, y_pred_binary)

# Afficher la matrice de confusion
plt.figure(figsize=(5,5))
plt.imshow(cm, interpolation='nearest', cmap=plt.cm.Blues)
plt.title('Matrice de confusion')
plt.colorbar()
plt.xticks([0, 1], ['Non frauduleuse', 'Frauduleuse'])
plt.yticks([0, 1], ['Non frauduleuse', 'Frauduleuse'])
plt.xlabel('Prédictions')
plt.ylabel('Réelles')
plt.show()

# Calculer la courbe ROC-AUC
fpr, tpr, _ = roc_curve(y_test, y_pred)
roc_auc = auc(fpr, tpr)


plt.figure()
plt.plot(fpr, tpr, color='darkorange', lw=2, label='Courbe ROC (AUC = %0.2f)' % roc_auc)
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlabel('Taux de faux positifs')
plt.ylabel('Taux de vrais positifs')
plt.title('Courbe ROC')
plt.legend(loc='lower right')
plt.show()


# Récupération des métriques
history_dict = history.history
loss_values = history_dict["loss"]
val_loss_values = history_dict["val_loss"]
accuracy_values = history_dict["accuracy"]
val_accuracy_values = history_dict["val_accuracy"]

epochs = range(1, len(loss_values) + 1)

# Courbe de la perte (Loss)
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(epochs, loss_values, "bo", label="Perte entraînement")
plt.plot(epochs, val_loss_values, "r-", label="Perte validation")
plt.title("Perte entraînement vs validation")
plt.xlabel("Époques")
plt.ylabel("Perte")
plt.legend()

# Courbe de la précision (Accuracy)
plt.subplot(1, 2, 2)
plt.plot(epochs, accuracy_values, "bo", label="Précision entraînement")
plt.plot(epochs, val_accuracy_values, "r-", label="Précision validation")
plt.title("Précision entraînement vs validation")
plt.xlabel("Époques")
plt.ylabel("Précision")
plt.legend()

plt.show()




# Fonction pour créer le modèle
def build_model(optimizer='adam', dropout_rate=0.3, neurons=64):
    model = Sequential([
        Dense(neurons, activation='relu', input_shape=(X_train.shape[1],)),
        Dropout(dropout_rate),
        Dense(neurons // 2, activation='relu'),
        Dropout(dropout_rate),
        Dense(1, activation='sigmoid')
    ])
    model.compile(optimizer=optimizer, loss='binary_crossentropy', metrics=['accuracy'])
    return model

# Utilisation du nouveau KerasClassifier
model = KerasClassifier(model=build_model, verbose=0)

# Définition des hyperparamètres à tester
param_grid = {
    'batch_size': [32, 64],
    'epochs': [10, 20],
    'optimizer': ['adam', 'sgd'],
    'dropout_rate': [0.2, 0.3, 0.5],
    'neurons': [128, 64, 32]
}


```

```python

# 🔹 Imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import confusion_matrix, classification_report, roc_curve, auc
from imblearn.over_sampling import SMOTE
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.utils import set_random_seed
from sklearn.base import BaseEstimator, ClassifierMixin
import shap
import lime
import lime.lime_tabular

# 🔹 Wrapper personnalisé compatible avec GridSearchCV
class KerasClassifierWrapper(BaseEstimator, ClassifierMixin):
    def __init__(self, learning_rate=0.001, dropout_rate=0.5, batch_size=1024, epochs=10):
        self.learning_rate = learning_rate
        self.dropout_rate = dropout_rate
        self.batch_size = batch_size
        self.epochs = epochs
        self.model_ = None

    def build_model(self):
        model = Sequential()
        model.add(Dense(64, input_dim=self.input_dim_, activation='relu'))
        model.add(Dropout(self.dropout_rate))
        model.add(Dense(32, activation='relu'))
        model.add(Dropout(self.dropout_rate))
        model.add(Dense(1, activation='sigmoid'))
        model.compile(optimizer=Adam(learning_rate=self.learning_rate), loss='binary_crossentropy', metrics=['accuracy'])
        return model

    def fit(self, X, y):
        self.input_dim_ = X.shape[1]
        self.model_ = self.build_model()
        self.model_.fit(X, y, batch_size=self.batch_size, epochs=self.epochs, verbose=0)
        return self

    def predict(self, X):
        return (self.model_.predict(X, verbose=0) > 0.5).astype("int32")

    def predict_proba(self, X):
        prob = self.model_.predict(X, verbose=0)
        return np.hstack([1 - prob, prob])


# 🔹 Chargement des données
file_path = "creditcard.csv"
df = pd.read_csv(file_path)

# 🔍 Exploration rapide
print("Dimensions :", df.shape)
print("Valeurs manquantes :\n", df.isnull().sum().sum())
print(df["Class"].value_counts(normalize=True))

# ⚙️ Prétraitement
scaler = StandardScaler()
df["Amount"] = scaler.fit_transform(df[["Amount"]])
df["Time"] = scaler.fit_transform(df[["Time"]])

X = df.drop("Class", axis=1)
y = df["Class"]

# 🔄 Rééquilibrage SMOTE
smote = SMOTE(random_state=42)
X_res, y_res = smote.fit_resample(X, y)
print("After SMOTE :", y_res.value_counts())

# ✅ Split des données
X_train, X_test, y_train, y_test = train_test_split(X_res, y_res, test_size=0.2, random_state=42)

# 🔍 GridSearchCV pour hyperparamètres
model_clf = KerasClassifierWrapper()
param_grid = {
    'learning_rate': [0.001, 0.0005],
    'dropout_rate': [0.3, 0.5],
    'batch_size': [1024],
    'epochs': [10]
}

grid = GridSearchCV(estimator=model_clf, param_grid=param_grid, cv=3, scoring='accuracy')
grid_result = grid.fit(X_train, y_train)

print("Best parameters:", grid_result.best_params_)
print("Best accuracy:", grid_result.best_score_)

# ✅ Évaluation du meilleur modèle
best_model = grid_result.best_estimator_
y_pred_prob = best_model.predict_proba(X_test)[:, 1]

# 📊 ROC Curve
fpr, tpr, thresholds = roc_curve(y_test, y_pred_prob)
roc_auc = auc(fpr, tpr)

plt.figure()
plt.plot(fpr, tpr, label=f"ROC curve (area = {roc_auc:.2f})")
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend()
plt.show()

# 📌 Matrice de confusion
y_pred = (y_pred_prob > 0.5).astype(int)
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title("Matrice de Confusion")
plt.xlabel("Prédit")
plt.ylabel("Réel")
plt.show()

# 🌈 Interprétabilité (SHAP)
explainer = shap.KernelExplainer(best_model.predict_proba, X_train[:100].values)
shap_values = explainer.shap_values(X_test[:5].values)
shap.summary_plot(shap_values, X_test[:5])
```

```python

# 🔹 Imports
import pandas as pd
import joblib

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import confusion_matrix, classification_report, roc_curve, auc
from imblearn.over_sampling import SMOTE
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.utils import set_random_seed
from sklearn.base import BaseEstimator, ClassifierMixin
import shap
import lime
import lime.lime_tabular
import keras_tuner as kt

# 🔹 Wrapper personnalisé compatible avec GridSearchCV
class KerasClassifierWrapper(BaseEstimator, ClassifierMixin):
    def __init__(self, learning_rate=0.001, dropout_rate=0.5, batch_size=1024, epochs=10):
        self.learning_rate = learning_rate
        self.dropout_rate = dropout_rate
        self.batch_size = batch_size
        self.epochs = epochs
        self.model_ = None

    def build_model(self):
        model = Sequential()
        model.add(Dense(64, input_dim=self.input_dim_, activation='relu'))
        model.add(Dropout(self.dropout_rate))
        model.add(Dense(32, activation='relu'))
        model.add(Dropout(self.dropout_rate))
        model.add(Dense(1, activation='sigmoid'))
        model.compile(optimizer=Adam(learning_rate=self.learning_rate), loss='binary_crossentropy', metrics=['accuracy'])
        return model

    def fit(self, X, y):
        self.input_dim_ = X.shape[1]
        self.model_ = self.build_model()
        self.model_.fit(X, y, batch_size=self.batch_size, epochs=self.epochs, verbose=0)
        return self

    def predict(self, X):
        return (self.model_.predict(X, verbose=0) > 0.5).astype("int32")

    def predict_proba(self, X):
        prob = self.model_.predict(X, verbose=0)
        return np.hstack([1 - prob, prob])


# 🔹 Chargement des données
file_path = "creditcard.csv"
df = pd.read_csv(file_path)

# 🔍 Exploration rapide
print("Dimensions :", df.shape)
print("Valeurs manquantes :\n", df.isnull().sum().sum())
print(df["Class"].value_counts(normalize=True))

# ⚙️ Prétraitement
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)  # ici X contient bien toutes les colonnes
joblib.dump(scaler, "scaler.pkl")   # ⬅️ sauver le scaler complet


X = df.drop("Class", axis=1)
y = df["Class"]

# 🔄 Rééquilibrage SMOTE
smote = SMOTE(random_state=42)
X_res, y_res = smote.fit_resample(X, y)
print("After SMOTE :", y_res.value_counts())

# ✅ Split des données
X_train, X_test, y_train, y_test = train_test_split(X_res, y_res, test_size=0.2, random_state=42)

# 🔍 GridSearchCV pour hyperparamètres
model_clf = KerasClassifierWrapper()
param_grid = {
    'learning_rate': [0.001, 0.0005],
    'dropout_rate': [0.3, 0.5],
    'batch_size': [1024],
    'epochs': [10]
}

grid = GridSearchCV(estimator=model_clf, param_grid=param_grid, cv=3, scoring='accuracy')
grid_result = grid.fit(X_train, y_train)

print("Best parameters:", grid_result.best_params_)
print("Best accuracy:", grid_result.best_score_)

# ✅ Évaluation du meilleur modèle
best_model = grid_result.best_estimator_
y_pred_prob = best_model.predict_proba(X_test)[:, 1]
y_pred = (y_pred_prob > 0.5).astype(int)

# 📊 ROC Curve
fpr, tpr, thresholds = roc_curve(y_test, y_pred_prob)
roc_auc = auc(fpr, tpr)

plt.figure()
plt.plot(fpr, tpr, label=f"ROC curve (area = {roc_auc:.2f})")
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend()
plt.show()

# 📌 Matrice de confusion
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title("Matrice de Confusion")
plt.xlabel("Prédit")
plt.ylabel("Réel")
plt.show()

# 📝 Rapport de classification
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))

# 🌈 Interprétabilité (SHAP)
explainer = shap.KernelExplainer(best_model.predict_proba, X_train[:100].values)
shap_values = explainer.shap_values(X_test[:5].values)
shap.summary_plot(shap_values, X_test[:5])

# 🌿 Interprétabilité (LIME)
explainer_lime = lime.lime_tabular.LimeTabularExplainer(
    training_data=np.array(X_train),
    feature_names=X.columns.tolist(),
    class_names=['Non Fraud', 'Fraud'],
    mode='classification'
)
i = 1
exp = explainer_lime.explain_instance(X_test.iloc[i].values, best_model.predict_proba, num_features=10)
exp.show_in_notebook(show_table=True)

# 🔧 Keras Tuner pour hyperparamètres

def model_builder(hp):
    model = Sequential()
    model.add(Dense(units=hp.Int('units', min_value=32, max_value=128, step=32), activation='relu', input_shape=(X_train.shape[1],)))
    model.add(Dropout(hp.Float('dropout', 0.2, 0.5, step=0.1)))
    model.add(Dense(1, activation='sigmoid'))
    model.compile(optimizer=Adam(hp.Choice('lr', [0.001, 0.0005])), loss='binary_crossentropy', metrics=['accuracy'])
    return model

tuner = kt.RandomSearch(
    model_builder,
    objective='val_accuracy',
    max_trials=3,
    executions_per_trial=1,
    directory='my_dir',
    project_name='fraud_tuning'
)

tuner.search(X_train, y_train, epochs=5, validation_split=0.2, verbose=0)
best_hp = tuner.get_best_hyperparameters(1)[0]
print("\nKeras Tuner Best Hyperparameters:")
print(f"Units: {best_hp.get('units')}, Dropout: {best_hp.get('dropout')}, Learning Rate: {best_hp.get('lr')}")

# 💾 Sauvegarde du modèle
best_model.model_.save("fraud_model.h5")

# 🚀 Déploiement - Étapes
print("\nÉtapes de déploiement en production :")
print("1. Exporter le modèle entraîné (.h5 ou .pb)")
print("2. Créer une API (ex: Flask, FastAPI) pour recevoir des données de transaction")
print("3. Charger le modèle et effectuer les prédictions en ligne")
print("4. Dockeriser l’application et la déployer sur un serveur cloud")
print("5. Mettre en place une supervision pour suivre les performances")

# 📈 Perspectives d'amélioration
print("\nPerspectives d’amélioration du modèle :")
print("- Collecter plus de données réelles et à jour")
print("- Tester d’autres architectures de réseau de neurones")
print("- Ajouter des fonctionnalités comportementales pour améliorer la détection")
print("- Implémenter une mise à jour continue du modèle en production")
```
